```python
from langgraph.graph import StateGraph, START, END
from langchain_ollama import ChatOllama
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model=ChatOllama(model="qwen3:8b")
```

```python
class blogstate(TypedDict):
    title : str
    outline : str
    content : str   
                
```

```python
def createoutline(state: blogstate) -> blogstate:

    title =  state['title']
    prompt = f'generate a detailed outline {title}'
    outline =  model.invoke(prompt).content
    state['outline'] = outline
    return state
      
 
```

```python
def create_blog(state: blogstate) -> blogstate:

    outline =  state['outline']
    prompt = f'generate a detailed blog {outline}'
    blog =  model.invoke(prompt).content
    state['content'] = blog
    return state
      
```

```python
graph = StateGraph(blogstate)

graph.add_node('createoutline',createoutline)
graph.add_node('create_blog',create_blog)
graph.add_edge(START, 'createoutline')
graph.add_edge('createoutline', 'create_blog')
graph.add_edge('create_blog', END)

workflow = graph.compile() 

```

```python
initial_state = {'title': 'How USA become the world only super power' }
final = workflow.invoke(initial_state)
print(final)
```
